# Análisis del e-commerce Olist

Olist es una plataforma brasileña de comercio electrónico que conecta vendedores con clientes de todo Brasil. Este proyecto analiza el comportamiento de las ventas, los pedidos, los pagos, las entregas, las reseñas y las categorías de productos a partir de los datos históricos disponibles.

El objetivo es comprender el desempeño del negocio, detectar oportunidades de mejora en la experiencia del cliente y obtener información útil para apoyar la toma de decisiones comerciales.

## Descripción de los datasets

El análisis integra los siguientes datasets del ecosistema de Olist:

- **Customers:** identifica a los clientes, sus pedidos y la ubicación registrada para cada compra.
- **Geolocation:** contiene las coordenadas geográficas asociadas a los códigos postales, ciudades y estados de Brasil.
- **Order items:** detalla los productos incluidos en cada pedido, junto con el vendedor, el precio y el costo de envío.
- **Payments:** registra los medios de pago utilizados, el valor pagado, el número de cuotas y la secuencia de pagos.
- **Reviews:** reúne las calificaciones y comentarios realizados por los clientes, además de las fechas de creación y respuesta.
- **Orders:** contiene el estado de cada pedido y las fechas de compra, aprobación, entrega al transportista, entrega al cliente y entrega estimada.
- **Products:** describe los productos vendidos, incluyendo categoría, dimensiones, peso y cantidad de fotografías.
- **Sellers:** identifica a los vendedores y su ubicación geográfica.
- **Category translation:** traduce los nombres de las categorías de productos del portugués al inglés para facilitar su interpretación.

En conjunto, estas tablas permiten analizar el recorrido completo de una compra: desde el cliente y el producto hasta el pago, la logística y la satisfacción final.

```python
import pandas as pd
from IPython.display import display
```

```python
ruta = "../Portafolio/archive/"

customers = pd.read_csv(ruta + "olist_customers_dataset.csv")
geolocation = pd.read_csv(ruta + "olist_geolocation_dataset.csv")
order_items = pd.read_csv(ruta + "olist_order_items_dataset.csv")
payments = pd.read_csv(ruta + "olist_order_payments_dataset.csv")
reviews = pd.read_csv(ruta + "olist_order_reviews_dataset.csv")
orders = pd.read_csv(ruta + "olist_orders_dataset.csv")
products = pd.read_csv(ruta + "olist_products_dataset.csv")
sellers = pd.read_csv(ruta + "olist_sellers_dataset.csv")
categories = pd.read_csv(ruta + "product_category_name_translation.csv")
```

### Interpretación de la primera visualización

El dataset **Customers** contiene **99.441 filas y 5 columnas**. Esto significa que se dispone de información de 99.441 registros de clientes, descritos mediante cinco variables: identificador del cliente, identificador único, código postal, ciudad y estado.

La función `head()` muestra las **primeras cinco filas** del dataset. Esta vista inicial permite comprobar que los datos se cargaron correctamente y reconocer la estructura y el tipo de información disponible antes de continuar con el análisis.

### **Generamos las tablas para identificar cuantas filas/columnas tiene cada tabla,sus nombres de columna y valores nulos**

```python
tablas = {
    "customers": customers,
    "geolocation": geolocation,
    "order_items": order_items,
    "payments": payments,
    "reviews": reviews,
    "orders": orders,
    "products": products,
    "sellers": sellers,
    "categories": categories
}

for nombre, df in tablas.items():
    print(f"\n{'='*50}")
    print(f"TABLA: {nombre}")
    print(f"Shape: {df.shape}")
    print(f"Columnas: {df.columns.tolist()}")
    print(f"Nulos:\n{df.isnull().sum()[df.isnull().sum() > 0]}")
```

### Interpretación de las tablas 

- **Customers:** tiene 99.441 filas y 5 columnas, sin valores nulos. La información básica de los clientes está completa.
- **Geolocation:** tiene 1.000.163 filas y 5 columnas, sin valores nulos. Los códigos postales cuentan con coordenadas, ciudad y estado registrados.
- **Order items:** tiene 112.650 filas y 7 columnas, sin valores nulos. Cada producto asociado a un pedido tiene información de vendedor, precio y envío.
- **Payments:** tiene 103.886 filas y 5 columnas, sin valores nulos. Todos los registros contienen información sobre el método de pago, cuotas y valor pagado.
- **Reviews:** tiene 99.224 filas y 7 columnas. Faltan 87.656 títulos y 58.247 comentarios, lo que indica que muchos clientes calificaron el pedido sin escribir comentarios. Estos campos pueden estar vacíos porque son opcionales.
- **Orders:** tiene 99.441 filas y 8 columnas. Hay valores nulos en las fechas de aprobación, entrega al transportista y entrega al cliente. Esto puede ocurrir en pedidos cancelados, no entregados o que todavía no habían completado esas etapas.
- **Products:** tiene 32.951 filas y 9 columnas. Hay 610 productos sin categoría, nombre, descripción o cantidad de fotografías, y 2 productos sin datos de peso y dimensiones. Estos valores deben considerarse al analizar las características de los productos.
- **Sellers:** tiene 3.095 filas y 4 columnas, sin valores nulos. La identificación y ubicación de los vendedores están completas.
- **Categories:** tiene 71 filas y 2 columnas, sin valores nulos. Todas las categorías tienen su correspondiente traducción al inglés.

En conclusión, la mayoría de las tablas tienen información completa. Los valores nulos encontrados se concentran principalmente en comentarios de clientes, fechas relacionadas con el proceso de entrega y características de algunos productos.

```python
duplicados = pd.DataFrame({
    "tabla": tablas.keys(),
    "filas_duplicadas": [tabla.duplicated().sum() for tabla in tablas.values()]
})

display(duplicados)
```
